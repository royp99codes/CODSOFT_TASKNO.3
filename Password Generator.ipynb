```python
class ToDo_List:
  def __init__(self, filename = "todo.txt"):
    self.filename = filename
    self.tasks = [] # Changed from self.task to self.tasks
    self.load_from_file()

  def load_from_file(self):
    try:
      with open(self.filename, "r") as file:
        for line in file:
          description, status = line.strip().split(",")
          task = {
            "description" : description,   "status" : status == "True"
          }
          self.tasks.append(task)
          # This print statement was inside the loop, typically it should be outside or removed.
          # For now, keeping it here but noting it might be better placed after the loop.
          print(f"Saved version loaded from {self.filename}")

    except FileNotFoundError:
      print("NO PREVIOUSLY SAVED VERSIONS FOUND...STARTING AN EMPTY TO DO LIST")

  def add_task(self,description):
    task = {
      "description" : description,   "status" :  False
    }
    self.tasks.append(task)
    self.save_to_file()
    print(f"task\"{description}\" added successfully to the list")

  def save_to_file(self):
    with open(self.filename, "w") as file:
      for task in self.tasks:
        # Fixed SyntaxError: unterminated string literal by using single quotes for dict keys
        line = f"{task['description']} , {task['status']}\n" # Added newline for proper file writing
        file.write(line)

  def display_tasks(self):
    print("----- TO DO LIST --------")

    # Fixed logic: pending tasks are not completed, completed tasks are.
    pending_tasks = [task for task in self.tasks if not task["status"]]
    completed_tasks = [task for task in self.tasks if task["status"]]

    if not pending_tasks and not completed_tasks:
      print("No Tasks available currently.....")
    else:
      print("------ PENDING TASKS -------")
      if pending_tasks:
        for i, task in enumerate(pending_tasks, 1):
          # Fixed SyntaxError: unterminated string literal
          print(f"{i}.{task['description']}  [TO-DO]")
      else:
        print("No pending tasks.")

      print("------ COMPLETED TASKS -------")
      if completed_tasks:
        for i, task in enumerate(completed_tasks, 1):
          # Fixed SyntaxError: unterminated string literal
          print(f"{i}.{task['description']}  [DONE]")
      else:
        print("No completed tasks.")

  def mark_as_done(self,task_index):
    pending_tasks = [task for task in self.tasks if not task["status"]]

    if 1 <= task_index <= len(pending_tasks):
      task_to_mark = pending_tasks[task_index - 1]
      # Find the actual task in self.tasks to update its status
      for task in self.tasks:
        if task['description'] == task_to_mark['description']:
          task['status'] = True
          break
      self.save_to_file()
      # Fixed SyntaxError: unterminated string literal
      print(f"task \"{task_to_mark['description']}\" marked as done")
    else:
      print("Invalid Task index")

  def remove_task(self, task_type, task_index):
    if task_type == 1: # Remove from Pending Tasks
      current_list = [task for task in self.tasks if not task["status"]]
      if 1 <= task_index <= len(current_list):
        task_to_remove = current_list[task_index - 1]
        self.tasks.remove(task_to_remove) # Remove the actual task object
        self.save_to_file()
        print(f"task \"{task_to_remove['description']}\" removed from list")
      else:
        print("Invalid Task index")

    elif task_type == 2: # Remove from Completed Tasks
      current_list = [task for task in self.tasks if task["status"]]
      if 1 <= task_index <= len(current_list):
        task_to_remove = current_list[task_index - 1]
        self.tasks.remove(task_to_remove) # Remove the actual task object
        self.save_to_file()
        print(f"task \"{task_to_remove['description']}\" removed from list")
      else:
        print("Invalid Task index")

    else:
      print("Invalid choice for task type")


def main(): # main function moved outside the class
  toDoList = ToDo_List() # Corrected class name instantiation

  while True:
    print("\n1. Display Tasks  \n2. Add new tasks \n3. Remove a task \n4. Mark task as done \n5. Quit")
    choice = input("\nPlease enter your choice (From 1 - 5): ")

    if choice == "1":
      toDoList.display_tasks()

    elif choice == "2":
      description = input("\nPlease enter task description: ")
      toDoList.add_task(description)

    elif choice == "3":
      toDoList.display_tasks()
      print("\nWant to Remove from: \n1. Pending Tasks \n2. Completed")
      task_type = int(input("Choose the type:"))
      toDoList.display_tasks() # Display tasks again for user to choose index
      task_index = int(input("Enter the task index to remove:")) # Missing input for task_index
      toDoList.remove_task(task_type, task_index) # Added missing task_type argument

    elif choice == "4":
      toDoList.display_tasks()
      task_index = int(input("\nEnter the task index to mark it as done:"))
      toDoList.mark_as_done(task_index)

    elif choice == "5":
      print("Exiting from programme..")
      break

    else:
      print("\n Please enter the valid choice...!")


if __name__ == "__main__": # Corrected if name to if __name__
  main()
```
